```python
# Imports
import os
from dotenv import load_dotenv
from huggingface_hub import login
from transformers import pipeline, BitsAndBytesConfig
import torch

load_dotenv(override=True)  # Load environment variables from .env file
```

```python
# Constants

LLAMA_MODEL = "meta-llama/Llama-3.2-3B-Instruct"
```

```python
audio_filename = "denver_extract.mp3"
```

```python
hf_token = os.environ.get("HF_TOKEN")
login(hf_token)

# Open the file

audio_file = open(audio_filename, "rb")
```

```python
pipe = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-medium.en",
    dtype=torch.float16,
    device="cuda",
    return_timestamps=True
)

result = pipe(audio_filename)
print(result)
```

```python
transcription = result["text"]
print(transcription)
```

```python
system_message = """
You produce minutes of meetings from transcripts, with summary, key discussion points,
takeaways and action items with owners, in markdown format without code blocks.
"""

user_prompt = f"""
Below is an extract transcript of a Denver council meeting.
Please write minutes in markdown without code blocks, including:
- a summary with attendees, location and date
- discussion points
- takeaways
- action items with owners

Transcription:
{transcription}
"""

messages = [
    {"role": "system", "content": system_message},
    {"role": "user", "content": user_prompt}
]

messages
```

```python
quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_quant_type="nf4"
)
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, TextStreamer

tokenizer = AutoTokenizer.from_pretrained(LLAMA_MODEL, use_fast=True)
tokenizer.pad_token = tokenizer.eos_token

inputs = tokenizer.apply_chat_template(messages, return_tensors="pt").to("cuda")

streamer = TextStreamer(tokenizer)

model = AutoModelForCausalLM.from_pretrained(LLAMA_MODEL, device_map="auto", quantization_config=quant_config)

output = model.generate(inputs, max_new_tokens=2000, streamer=streamer)
```

```python
from IPython.display import Markdown

r = tokenizer.decode(output[0])
display(Markdown(r))
```
